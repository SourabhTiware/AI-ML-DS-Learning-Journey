### Lec 04: Important Plot methods

```python
# Oscar Winner Movies[cite: 1]
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M[cite: 1]
```

```python
plt.plot(years,oscar_revenue)
plt.title("Oscar Winner Movies")
plt.xlabel("Year's")
plt.ylabel("Revenue in ($M)" )
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a DataFrame
df = pd.DataFrame({'years': years, 'revenue': oscar_revenue})
# df
# Pandas plot method accepts these directly
df.plot(x='years', y='revenue', title="Oscar Winner Movies Revenue", xlabel="Years", ylabel="Revenue ($M)")
plt.show()
```

```python

def plot_with_details(x, y, title="", xlabel="", ylabel="", marker=""):
    plt.plot(x, y, marker=marker)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()

# Now you can pass them all into your custom method
plot_with_details(
    years, 
    oscar_revenue, 
    title="Oscar Winner Movies Revenue", 
    xlabel="Years", 
    ylabel="Revenue ($M)", 
    marker="o"
)
```

### Lec 05: Multiple Data set on line plot

```python
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years, oscar_revenue, label="Oscar")
plt.plot(years, non_oscar_revenue, label="Non Oscar")

plt.title("Oscar Movies VS non oscar movies")
plt.xlabel("Year's")
plt.ylabel("Revenue in ($M")
plt.legend()

plt.show()
```

#### why both plot line chart on same chart?
#### why is not create sepearte 2 charts?? 

Both lines appear on the same chart because of how Matplotlib's state machine works by default:

The Single Canvas Rule: When you call plt.plot(), Matplotlib looks at the current active chart (axes). Every time you call plt.plot() again without clearing or creating a new figure, Matplotlib simply draws the new line right on top of the existing canvas.

Designed for Comparison: This behavior is intentional—it allows you to overlay multiple data series (like comparing "Oscar" vs. "Non Oscar") on the exact same coordinate system so you can easily compare their trends against the same X and Y axes.

How to create separate charts instead:
If you want two distinct, separate charts instead of combining them into one, you need to explicitly tell Matplotlib to create a new figure or window using plt.figure() before your second plot:


```python
import matplotlib.pyplot as plt

# --- Chart 1: Oscar Movies ---
plt.figure() # Creates a new window/canvas for the first chart
plt.plot(years, oscar_revenue, label="Oscar", color="blue")
plt.title("Oscar Winner Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
plt.legend()
plt.show() # Displays and clears the first chart

# --- Chart 2: Non-Oscar Movies ---
plt.figure() # Creates a separate new window/canvas for the second chart
plt.plot(years, non_oscar_revenue, label="Non Oscar", color="orange")
plt.title("Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
plt.legend()
plt.show() # Displays the second chart independently
```

# Lec 10: Multiple Dataset on Bar Chart

```python
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]

import numpy as np

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="oscar")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non oscar")

plt.xlabel("years")
plt.ylabel("Revenue (in $M")
plt.title("Yearly revenue of oscar vs Non-oscar movies")
plt.legend()
plt.tight_layout()

plt.xticks(x, years) # to show x-axis labels
```
