## Beautiful Soap

```python
from bs4 import BeautifulSoup

with open("scrapped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
# all_h3 = soup.find_all("h3")
# price = soup.find_all("p")

# books = []
# prices = []

# all_book =[]

# for h3 in all_h3:
#     book = h3.find_next("a").get_text(strip=True)
#     # print(book)
#     books.append(book)


# # for b in books:
# #     print(b)

# for p in price:
#     div = p.find_next("div")
#     if div and div.select(".price_color"):
#         price = div.select_one(".price_color").get_text(strip=True)[2:]
#         # print(price)
#         prices.append(price)

# # for p in prices:
# #     print(p)


all_book = []

# Select each product container on the page
products = soup.select(".product_pod")

for product in products:
    book = product.h3.a.get_text(strip=True)
    price = product.select_one(".price_color").get_text(strip=True)[2:]
    
    all_book.append([book, price])

# Output format: [['A Light in the ...', '51.77'], ['Tipping the ...', '53.74'], ...]
print(all_book)


```

```python
book_col = pd.DataFrame(all_book, columns=["Name","price"])
```

```python
book_col.to_csv("cleaned_data/data.csv", index=False)
```
